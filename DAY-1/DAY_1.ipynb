```python
# Student Performance Explorer

# import pandas
import pandas as pd

# load csv file
df = pd.read_csv("student_performance.csv")

# create total score column
df["total_score"] = (
    df["math_score"] +
    df["programming_score"]
)

# project title
print("Student Performance Explorer")

# dataset overview
print("\nDataset Overview")

# total students
print("Total students:", len(df))

# total columns
print("Total columns:", len(df.columns))

# unique departments
print("Unique departments:", df["department"].nunique())

# department wise analysis
print("\nDepartment Wise Analysis")

# count students in each department
dept_count = df["department"].value_counts()

# print department counts
for dept, count in dept_count.items():
    print(dept, ":", count)

# score analysis
print("\nScore Analysis")

# highest math score
print("Highest math score:", df["math_score"].max())

# lowest math score
print("Lowest math score:", df["math_score"].min())

# average math score
print("Average math score:", df["math_score"].mean())

# average programming score
print("Average programming score:", df["programming_score"].mean())

# average attendance
print("Average attendance:", df["attendance_percentage"].mean())

# gender wise analysis
print("\nGender Wise Analysis")

# average total score by gender
gender_avg = df.groupby("gender")["total_score"].mean()

# print gender averages
for gender, avg in gender_avg.items():
    print(gender, "average total score:", avg)

# top performers
print("\nTop 5 Performers")

# sort by total score
top_students = df.sort_values(
    by="total_score",
    ascending=False
).head(5)

# print top students
for index, row in top_students.iterrows():
    print("Name:", row["name"])
    print("Department:", row["department"])
    print("Total Score:", row["total_score"])
    print()

# attendance analysis
print("\nAttendance Analysis")

# students below 75 attendance
low_attendance = df[
    df["attendance_percentage"] < 75
]

# check low attendance students
if len(low_attendance) > 0:

    for index, row in low_attendance.iterrows():
        print("Name:", row["name"])
        print("Department:", row["department"])
        print("Attendance:", row["attendance_percentage"])
        print()

else:
    print("All students have attendance above 75%")

# practice questions
print("\nPractice Questions")

# 1 electronics department students
electronics_count = len(
    df[df["department"] == "Electronics"]
)

print("1. Electronics students:", electronics_count)

# 2 average attendance of female students
female_attendance = df[
    df["gender"] == "Female"
]["attendance_percentage"].mean()

print("2. Average female attendance:", female_attendance)

# 3 lowest programming score student
lowest_prog_student = df.loc[
    df["programming_score"].idxmin()
]

print(
    "3. Lowest programming score student:",
    lowest_prog_student["name"]
)

# 4 students scored above 80 in math
math_above_80 = len(
    df[df["math_score"] > 80]
)

print("4. Students above 80 in math:", math_above_80)

# 5 average total score for attendance above 90
avg_total_above_90 = df[
    df["attendance_percentage"] > 90
]["total_score"].mean()

print(
    "5. Average total score above 90 attendance:",
    avg_total_above_90
)
```
